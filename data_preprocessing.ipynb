```python
# converting numeric columns since the dataset uses "?" for missing values
df.replace('?',np.nan,inplace=True)
numeric_colums =[
    "Global_active_power",
    "Global_reactive_power",
    "Voltage",
    "Global_intensity",
    "Sub_metering_1",
    "Sub_metering_2",
    "Sub_metering_3"
]
df[numeric_colums] = df[numeric_colums].astype(float)
```

```python
df["Date"].nunique()
df["Time"].nunique() # 1440 unique timestamps
```

```python
#removing missing values
df = df.dropna()
df.isnull().sum()
```

```python
#converting object data types to float
df[numeric_colums] = df[numeric_colums].astype(float)
```

```python
# combining data and time instead of keeping them as strings
df["Datetime"]= pd.to_datetime(df["Date"]+ " " +df["Time"])
```

```python
# extracting the hour, day and month 
df["Hour"]= df["Datetime"].dt.hour
df["Day"]= df["Datetime"].dt.day
df["Month"] = df["Datetime"].dt.month
```

```python
from sklearn.preprocessing import StandardScaler
scaler=StandardScaler()
scaled_features = scaler.fit_transform(df[numeric_colums])
df_scaled = pd.DataFrame(scaled_features, columns=numeric_colums)
```

```python
# removing original date and time features
df.drop(["Date","Time"],axis=1,inplace=True)
```
